```python


from reportlab.pdfgen import canvas

def create_pdf(file_name, text):
    c = canvas.Canvas(file_name)
    text_object = c.beginText(40, 800)
    for line in text.split('\n'):
        text_object.textLine(line)
    c.drawText(text_object)
    c.save()

policy1 = """Policy on Trade Volume:
- Any trade over 2000 units should be escalated to compliance for review.

Policy on Insider Trading:
- Trading by employees based on unpublished financial results is strictly prohibited."""

policy2 = """Policy on Trade Timing:
- Trades within 3 days before earnings release must be reported to compliance.

Policy on Approval:
- All trades in sensitive instruments must be pre-approved."""

create_pdf("pdf_files/policy1.pdf", policy1)
create_pdf("pdf_files/policy2.pdf", policy2)
```
